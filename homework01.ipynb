```python
# 📦 **Step 2: Import Required Libraries**
# 
# Let's import all the libraries we need for this project:

import os  # For accessing environment variables (like API keys)
from pydantic_ai import Agent  # Main AI agent framework
from pydantic_ai.models.openai import OpenAIChatModel  # AI model interface
from pydantic_ai.providers.openrouter import OpenRouterProvider  # Provider for accessing AI models
import requests  # For making HTTP requests to APIs
import json  # For handling JSON data (news articles come in JSON format)

# This is needed for Jupyter notebooks to handle async operations properly
import nest_asyncio
nest_asyncio.apply()  # Allows nested event loops in Jupyter
```

```python
# 🔍 **Step 3: News Retrieval Function**
# 
# This function fetches news articles from the Serper API (Google Search API).
# It takes a keyword and returns a list of news articles related to that keyword.
# 
# **What this function does:**
# 1. Makes a POST request to Serper's news API
# 2. Searches for news in Hong Kong (location: "Hong Kong")
# 3. Returns news from the past week (tbs: "qdr:w")
# 4. Returns results in Traditional Chinese (hl: "zh-tw")
def get_news(keyword: str) -> list[dict]:
    """
    Fetch news articles from Serper API based on a search keyword.
    
    Args:
        keyword (str): The search term to find news about
        
    Returns:
        list[dict]: A list of news articles, each containing title, snippet, etc.
        
    Example:
        news_articles = get_news("iPhone")
        print(f"Found {len(news_articles)} articles")
    """
    # Serper API endpoint for news search
    url = "https://google.serper.dev/news"

    # Prepare the search parameters
    # q: search query, location: geographic location, gl: country code
    # hl: language, tbs: time-based search (qdr:w = past week)
    payload = json.dumps({
        "q": keyword,                    # What to search for
        "location": "Hong Kong",         # Search location
        "gl": "hk",                      # Country code (Hong Kong)
        "hl": "zh-tw",                   # Language (Traditional Chinese)
        "tbs": "qdr:w"                   # Time range (past week)
    })
    
    # Set up headers for the API request
    headers = {
        'X-API-KEY': os.getenv('SERPER_API_KEY'),  # Your API key from environment
        'Content-Type': 'application/json'          # Tell API we're sending JSON
    }

    # Make the API request
    response = requests.request("POST", url, headers=headers, data=payload)

    # Convert the response to Python dictionary
    parsed = response.json()
    
    # Return just the news articles (the API returns other metadata too)
    return parsed['news']
```

```python
# 📝 **Step 4: News Summarization Function**
# 
# This function takes the raw news data and formats it into a readable summary.
# It extracts the title and snippet from each news article and formats them nicely.
def summarize_news(news: list[dict]) -> str:
    """
    Format news articles into a readable summary string.
    
    Args:
        news (list[dict]): List of news articles from get_news() function
        
    Returns:
        str: Formatted summary with titles and snippets
        
    Example:
        articles = get_news("weather")
        summary = summarize_news(articles)
        print(summary)
    """
    # Start with an empty string to build our summary
    summary = ""
    
    # Loop through each news article in the list
    for article in news:  # 'article' is clearer than 'n'
        # Format each article with title (as header) and snippet (as content)
        # f-string allows us to insert variables into the string
        summary += f"###{article['title']}\n{article['snippet']}\n"
    
    return summary
```

```python
# 🧪 **Step 5: Test Our Functions**
# 
# Let's test our functions to make sure they work correctly!
# We'll search for news about "米娜" (a typhoon name) and see what we get.

# This line does two things:
# 1. get_news("米娜") - fetches news articles about "米娜"
# 2. summarize_news(...) - formats those articles into a readable summary
summary = summarize_news(get_news("米娜"))

# Print the results so we can see what we got
print(summary)
```

```python
# 🤖 **Step 6: Set Up AI Agent for Final Summarization**
# 
# Now we'll create an AI agent that can take all our news summaries and create
# a comprehensive, professional summary. This agent will use a language model
# to understand and synthesize the information.

# Define the instructions for our AI agent
"""
My version
[[Instruction]]
You are a news summarizer.
You will be given a list of news organized by keyword.
You will be given today's date so mind the date in the news when summarizing the news. News might be dated in the past but mentioned as today's news.
Always mind the date when summarizing the news.
You need to summarize the news in a professional yet informative manner. Do not miss any important information.
Please always answer in Traditional Chinese and Markdown format.
"""

system_prompt = """
You are a professional news summarizer. Your task is to create comprehensive, accurate summaries of news articles.

## Your Role:
- Analyze news articles organized by keywords
- Create professional, informative summaries
- Always respond in Traditional Chinese using Markdown format

## Important Guidelines:

### 📅 Date Awareness:
- Pay close attention to dates in the news
- Consider the context of when events occurred vs. when they're being reported
- Use the provided current date as reference point

### 📝 Summary Requirements:
- Include all important information from the articles
- Organize content logically by topic/keyword
- Use clear headings and bullet points for readability
- Maintain professional tone while being informative

### 🔍 Quality Standards:
- Verify information accuracy
- Avoid speculation or personal opinions
- Focus on facts and key developments
- Ensure comprehensive coverage of all provided news

## Output Format:
Structure your response with:
- Clear section headings for each keyword/topic
- Bullet points for key information
- Proper Markdown formatting
- Traditional Chinese language throughout
"""

# Create the AI model - we're using Google's Gemini model through OpenRouter
# This is a fast, efficient model good for summarization tasks
model = OpenAIChatModel(
    "google/gemini-2.5-flash-lite",  # Model name
    provider=OpenRouterProvider(api_key=os.getenv('OPENROUTER_API_KEY')),  # API provider
)

# Create our AI agent with the model and instructions
agent = Agent(
    model,                    # The AI model to use
    system_prompt=system_prompt  # The instructions we defined above
)
```

```python
# 📊 **Step 7: Process Multiple Keywords**
# 
# Now we'll search for news about multiple topics and prepare everything
# for our AI agent to create a comprehensive summary.

# Define the topics we want to search for
keywords = ["米娜", "iphone 17 pro", "港股"]

# Create a dictionary to store summaries for each keyword
summary_list = dict()

# Loop through each keyword and get news summaries
for keyword in keywords:
    print(f"🔍 Searching for news about: {keyword}")
    
    # Get news articles for this keyword
    news_articles = get_news(keyword)
    
    # Format the articles into a summary
    summary_list[keyword] = summarize_news(news_articles)
    
    print(f"✅ Found {len(news_articles)} articles for '{keyword}'")

# Get today's date for the AI agent (important for context)
import datetime
today = datetime.date.today().isoformat()

# Convert our summaries to JSON format for the AI agent
# ensure_ascii=False allows Chinese characters to display properly
news_json = json.dumps(summary_list, ensure_ascii=False)

# Create the prompt we'll send to our AI agent
user_prompt = f"""
[[Date]]
{today}

[[News]]
{news_json}
"""
```

```python
# 👀 **Step 8: Preview the Data**
# 
# Let's take a look at what we're sending to our AI agent.
# This helps us understand the format and content before processing.

print("📋 Here's what we're sending to the AI agent:")
print("=" * 50)
print(user_prompt)
```

```python
# 🚀 **Step 9: Generate Final Summary with AI**
# 
# Now we'll send all our news data to the AI agent to create a comprehensive,
# professional summary. The AI will analyze all the news and create a coherent report.

# Note: We use 'await' because the AI agent runs asynchronously
# This means it can take a few seconds to process all the information
print("🤖 AI agent is processing the news...")
final_summary = await agent.run(user_prompt)
```

```python
# 📄 **Step 10: Display the Final Summary**
# 
# Finally, let's display our AI-generated summary in a nice, formatted way.
# IPython's Markdown display makes it look professional and easy to read.

from IPython.display import Markdown

# Display the final summary with proper formatting
print("📰 Here's your comprehensive news summary:")
print("=" * 50)
Markdown(final_summary.output)
```
