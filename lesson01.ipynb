````python
import os
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openrouter import OpenRouterProvider


model = OpenAIChatModel(
    "google/gemini-2.5-flash-lite",
    provider=OpenRouterProvider(api_key=os.getenv('OPENROUTER_API_KEY')),
)
agent = Agent(model)
question = """
請提供以下新聞的撮要：


「世界末日」《我所看見的未來》7.5地震預言，愈說得多愈易「顯化」？對未知的焦慮情緒可如何應對？｜Winnie《心靈治療師「為你娘」》
龍樹諒預言7.5大地震！老闆擔心世界末日決定放假1天︰好好陪家人
7月5日末日預言倒數2天？美國資深地震專家「1句話」打臉
7/5 預言即將來臨？盤點 4 個歷史上的末日預言，從經驗中學習如何安心準備，應對防災與緊急避難
7／5大地震預言倒數！盤點那些年「沒成真的末日預言」 2012全球恐慌
7月5日末日預言成真？專家曝：引發海嘯規模8地震「有1前兆」
7/5世界末日預言瘋傳！台專家揭重大發現 規模8強震「1前兆曝光」
地震預言先別慌！日旅達人猜測7月5日最大新聞是這條 「不是過了就沒事」
7/5末日預言波及台灣？地震專家曝2條海溝如「高速公路」 恐吸海嘯直撲
懶人包／7月5日大地震倒數中？細數9次翻船末日世界預言
"""
result = await agent.run(question)
print(result.output)
````

````python
import requests
import json

url = "https://google.serper.dev/news"

payload = json.dumps({
  "q": "波鞋",
  "location": "Hong Kong",
  "gl": "hk",
  "hl": "zh-tw",
  "tbs": "qdr:w"
})
headers = {
  'X-API-KEY': os.getenv('SERPER_API_KEY'),
  'Content-Type': 'application/json'
}

response = requests.request("POST", url, headers=headers, data=payload)

parsed = response.json()

for n in parsed['news']: # for each elements in parsed['news'], do the following
  print(n['title'])
  print("    "+n['snippet'])
````

````python
news = parsed['news']
for n in news: # Control structure
    print(n['title'])
    if ('利物浦' in n['title']) or ("離世" in n['title']):
      # print("******" + n['snippet'] + "******")
      print(f"提要：{n['snippet']}")
    else:
      print(n['snippet'])
    print("\n")
print("This is the instruction")
````
